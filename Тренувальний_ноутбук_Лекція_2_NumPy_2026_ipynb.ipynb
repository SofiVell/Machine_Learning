## 13. Лінійна алгебра

### Матричний добуток

```python
a = np.array([
    [1, 2, 3],
    [0, 1, 2]
])

b = np.array([
    [ 1, 2],
    [ 0, 4],
    [-1, 0]
])

c = a @ b
print(c)
```

### Діагональ, транспонування та визначник

```python
a = np.array([
    [ 1.,  0., 3.],
    [-1., -1., 2.],
    [ 4.,  7., 2.]
])

print("diag:", np.diag(a))
print("\nA.T =\n", a.T)

det = np.linalg.det(a)
print("\ndet(A) =", det)
print("det близький до -25:", np.isclose(det, -25.0))
```

### Розв’язання системи `Ax = b`

```python
a = np.array([
    [1.,  2.,  5.],
    [1., -1.,  3.],
    [3., -6., -1.]
])

b = np.array([-9., 2., 25.])

x = np.linalg.solve(a, b)

print("x =", x)
print("Перевірка A @ x ≈ b:", np.allclose(a @ x, b))
```

Для розв’язання системи не потрібно явно обчислювати `A⁻¹`.  
У числових обчисленнях перевага надається `np.linalg.solve(A, b)`.


### Метод найменших квадратів

```python
x = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.],
    [1., 4.]
])

y = np.array([2.1, 3.9, 6.2, 8.1])

beta, residuals, rank, s = np.linalg.lstsq(x, y, rcond=None)

print("beta:", beta)
print("residuals:", residuals)
print("rank:", rank)
print("singular values:", s)
```

### Власні значення та власні вектори

```python
a = np.array([
    [2., 1.],
    [1., 2.]
])

values, vectors = np.linalg.eigh(a)

print("Власні значення:", values)
print("Власні вектори:\n", vectors)
```

### SVD

```python
a = np.array([
    [3., 1.],
    [1., 3.],
    [1., 1.]
])

U, s, Vt = np.linalg.svd(a, full_matrices=False)

print("U =\n", U)
print("\ns =", s)
print("\nVt =\n", Vt)
```

## 18. Узагальнюючий приклад: підготовка даних і лінійний прогноз

```python
# 1. Вхідні дані: два фактори та цільова змінна
x = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(x).all(axis=1) & np.isfinite(y)
x = x[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = x.mean(axis=0)
sigma = x.std(axis=0)
x_std = (x - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
x_design = np.column_stack([
    np.ones(x_std.shape[0]),
    x_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(x_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = x_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

```python
# 1. Створіть матриця X форми (5, 4) із випадкових чисел
rng = np.random.default_rng()

X = rng.normal(size=(5, 4))

print("\n1.Створіть матриця X форми (5, 4) із випадкових чисел:")

print("X =\n", X)


# 2. Обчисліть середнє та стандартне відхилення кожної ознаки
mean = X.mean(axis=0)
std = X.std(axis=0)

print("\n2.Обчисліть середнє та стандартне відхилення кожної ознаки:")

print("\nСереднє кожної ознаки:")
print(mean)

print("\nСтандартне відхилення кожної ознаки:")
print(std)


# 3. Стандартизуйте матрицю через broadcasting
X_std = (X - mean) / std

print("\n3.Стандартизуйте матрицю через broadcasting:")

print("\nСтандартизована матриця:")
print(X_std)


# 4. Виберіть рядки, у яких значення першої ознаки додатне
X_positive = X[X[:, 0] > 0]

print("\n4.Виберіть рядки, у яких значення першої ознаки додатне:")

print("\nРядки, де перша ознака додатна:")
print(X_positive)


# 5. Додайте до матриці новий стовпець одиниць
X_with_bias = np.column_stack([
    np.ones(X_std.shape[0]),
    X_std
])

print("\n5.Додайте до матриці новий стовпець одиниць:")

print("\nМатриця зі стовпцем одиниць:")
print(X_with_bias)


# 6. Створіть вектор ваг і обчисліть прогноз X @ w
w = np.array([0.5, -0.2, 0.3, 0.1])

y_pred = X_std @ w

print("\n6.Створіть вектор ваг і обчисліть прогноз X @ w:")

print("\nВектор ваг:")
print(w)

print("\nПрогноз y_pred:")
print(y_pred)


# 7. Для заданого y_true обчисліть MSE
y_true = np.array([1.2, 0.8, -0.5, 1.5, 0.3])

mse = np.mean((y_true - y_pred) ** 2)

print("\n7.Для заданого y_true обчисліть MSE:")

print("\nMSE =", mse)


# 8. Перевірте, чи немає в масивах NaN або inf

print("\n8.Перевірте, чи немає в масивах NaN або inf:")

print("\nПеревірка X:")
print("Усі значення скінченні:", np.isfinite(X).all())

print("\nПеревірка X_std:")
print("Усі значення скінченні:", np.isfinite(X_std).all())

print("\nПеревірка y_pred:")
print("Усі значення скінченні:", np.isfinite(y_pred).all())

print("\nПеревірка y_true:")
print("Усі значення скінченні:", np.isfinite(y_true).all())
```

## Контрольні запитання

1. Чим `ndarray` відрізняється від списку Python?

ndarray — масив NumPy, у якому всі елементи зазвичай мають один тип даних (dtype). Він оптимізований для швидких числових обчислень і підтримує векторизацію та багатовимірні операції.

Список Python може містити елементи різних типів і не підтримує математичні операції над усіма елементами одночасно.

2. Що означають `shape`, `ndim`, `size`, `dtype`, `itemsize`?

shape — розмір масиву по кожній осі, наприклад (3, 4)

ndim — кількість вимірів (осей)

size — загальна кількість елементів

dtype — тип даних елементів

itemsize — кількість байтів, яку займає один елемент

3. У чому різниця між `np.arange` і `np.linspace`?

np.arange(start, stop, step) створює значення із заданим кроком.

np.linspace(start, stop, num) створює задану кількість рівновіддалених значень.

4. Чим `view` відрізняється від `copy`?

view використовує ті самі дані, що й вихідний масив. Тому зміна view може змінити початковий масив.

copy створює незалежну копію даних. Зміна копії не впливає на початковий масив.

5. Що таке векторизація?

Векторизація — це виконання операцій над цілими масивами без явних циклів Python.

Наприклад, a ** 2 підносить до квадрата всі елементи масиву одночасно.Це робить код коротшим і зазвичай швидшим.

6. Які правила broadcasting?

При операціях над масивами NumPy порівнює їхні розміри, починаючи з останньої осі. Два розміри сумісні, якщо вони однакові або один із них дорівнює 1. Якщо масив має менше вимірів, NumPy може умовно додати відсутні осі.

7. Як трактувати `axis=0` і `axis=1`?

axis=0 — операція виконується по стовпцях, результат для кожного стовпця.

axis=1 — операція виконується по рядках, результат для кожного рядка.

8. Чим відрізняються `*` та `@`?

`*` виконує поелементний добуток.

`@` виконує матричний добуток або скалярний добуток векторів.

9. Чому для `Ax=b` краще використовувати `np.linalg.solve`?

np.linalg.solve(A, b) спеціально призначений для розв'язання системи лінійних рівнянь.

10. Чому для нового коду рекомендовано `np.random.default_rng()`?

np.random.default_rng() використовує сучасний API для генерації випадкових чисел.
